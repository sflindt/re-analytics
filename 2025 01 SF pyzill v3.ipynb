```python
## DEFINE BOUNDING BOX FUNCTION
from geopy import Point
from geopy.distance import geodesic

def get_bounding_box(center_lat, center_long, target_area=100):
    """
    Calculate the bounding box coordinates for a city based on the target area.

    Parameters:
        center_lat (float): Latitude of the city center.
        center_long (float): Longitude of the city center.
        city_name (str): Name of the city.
        target_area (float): Area in square miles for the bounding box (default is 100).

    Returns:
        tuple: Bounding box coordinates (ne_lat, ne_long, sw_lat, sw_long).
    """
    # Calculate half of the side length for the bounding box
    half_side_length = (target_area ** 0.5) / 2

    # Calculate the four bounding box points (N, S, E, W) around the center point
    north_point = geodesic(miles=half_side_length).destination(Point(center_lat, center_long), 0)  # 0° North
    south_point = geodesic(miles=half_side_length).destination(Point(center_lat, center_long), 180)  # 180° South
    east_point = geodesic(miles=half_side_length).destination(Point(center_lat, center_long), 90)  # 90° East
    west_point = geodesic(miles=half_side_length).destination(Point(center_lat, center_long), 270)  # 270° West

    # Extract lat/long for each bounding box corner
    ne_lat, ne_long = north_point.latitude, east_point.longitude
    sw_lat, sw_long = south_point.latitude, west_point.longitude

    # Print or return the coordinates
    print(f"Bounding box for {city_name}:")
    print(f"NE: ({ne_lat}, {ne_long})")
    print(f"SW: ({sw_lat}, {sw_long})")

    return ne_lat, ne_long, sw_lat, sw_long

# # Example usage:
# city_name = "Salt Lake City"
# center_lat = 40.7608
# center_long = -111.8910

# ne_lat, ne_long, sw_lat, sw_long = get_bounding_box(center_lat, center_long)

```

```python
##CODE TO PULL UNIQUE LISTINGS FROM MULTIPLE CITIES APPEND CITY TAG
import csv
from geopy import Point
from geopy.distance import geodesic
from datetime import datetime


# Get the current timestamp for unique filenames
current_time = datetime.now().strftime("%Y_%m%d_%H%M")


# Define the parameters and list of cities with their center coordinates and target area
target_listings = 500
zoom_value = 1
cities = [
    {"name": "Millcreek", "lat": 40.682460, "long": -111.846810, "target_area": 9},
    {"name": "Avenues", "lat": 40.780370, "long": -111.875450, "target_area": 4},
    {"name": "Salt Lake City", "lat": 40.7608, "long": -111.8910, "target_area": 100},
    # {"name": "Issaquah", "lat": 47.5301, "long": -122.0329, "target_area": 10},
    # {"name": "Kirkland", "lat": 47.716522456444096, "long": -122.20590271158834, "target_area": 40},
    # {"name": "Draper, UT", "lat": 40.5247, "long": -122.0329, "target_area": 16},
    # {"name": "Salisbury", "lat": 35.66787348371489, "long": -80.47244053847189, "target_area": 25},  # Salisbury, NC
    {"name": "Provo", "lat": 40.2338, "long": -111.6585, "target_area": 40},
    {"name": "Seattle", "lat": 47.6062, "long": -122.3321, "target_area": 25}

]

# # Generate a filename with the city names
# city_names = "_".join(city["name"].replace(" ", "") for city in cities)  # Remove spaces and join names with underscores
# file_name = f"{current_time}_{city_names}_MF_data.csv"

# Generate a simplified filename with a timestamp and city count
city_count = len(cities)  # Count the number of cities
file_name = f"{current_time}_{city_count}_cities_MF_data.csv"



# Initialize tracking variables for unique IDs and combined results
unique_ids = set()
combined_results = []

# Define fields for CSV output
fields = [
    'zpid', 'rawHomeStatusCd', 'homeStatusForHDP', 'homeStatus',
    'homeType', 'address', 'addressZipcode', 'addressCity', 'addressState',
    'latitude', 'longitude', 'detailUrl', 'unformattedPrice', 'beds', 'baths',
    'area', 'zestimate', 'rentZestimate', 'taxAssessedValue', 'timeOnZillow',
    'city_tag'  # New column for city tag
]

# Function to calculate bounding box
def get_bounding_box(center_lat, center_long, target_area):
    half_side_length = (target_area ** 0.5) / 2
    north_point = geodesic(miles=half_side_length).destination(Point(center_lat, center_long), 0)
    south_point = geodesic(miles=half_side_length).destination(Point(center_lat, center_long), 180)
    east_point = geodesic(miles=half_side_length).destination(Point(center_lat, center_long), 90)
    west_point = geodesic(miles=half_side_length).destination(Point(center_lat, center_long), 270)
    return north_point.latitude, east_point.longitude, south_point.latitude, west_point.longitude

# Loop through each city
for city in cities:
    city_name = city["name"]
    center_lat = city["lat"]
    center_long = city["long"]
    target_area = city["target_area"]  # Get the city-specific target area

    # Calculate bounding box for the city
    ne_lat, ne_long, sw_lat, sw_long = get_bounding_box(center_lat, center_long, target_area)

    # Initialize page for pagination
    page = 1

    # Loop to fetch unique listings for the city
    while len(combined_results) < target_listings:
        # Fetch data for the current page
        data = pyzill.search_multi_family(page, ne_lat, ne_long, sw_lat, sw_long, zoom_value)
        page_listings = data.get("listResults", [])

        # Track the number of unique listings found on this page
        initial_count = len(combined_results)

        # Filter and add only unique listings to combined_results
        for listing in page_listings:
            zpid = listing.get("zpid")
            if zpid not in unique_ids:
                unique_ids.add(zpid)  # Add zpid to the set to track it as seen
                listing["city_tag"] = city_name  # Add city tag to the listing
                combined_results.append(listing)  # Add listing to the combined results

        # Check if any new unique listings were added in this iteration
        if len(combined_results) == initial_count:
            print(f"No new unique listings found on page {page} for {city_name}. Stopping loop.")
            break  # Exit the loop if no new unique listings were added

        print(f"City: {city_name}, Page {page} processed. Total unique listings so far: {len(combined_results)}")

        # Increment to the next page
        page += 1

        # Optional: Stop if no more data is returned
        if not page_listings:
            print(f"No more listings available on page {page} for {city_name}. Stopping loop.")
            break

# Write the combined unique results to a single CSV file
with open(file_name, "w", newline="") as csv_file:
    writer = csv.DictWriter(csv_file, fieldnames=fields)
    writer.writeheader()

    for listing in combined_results:
        hdp_data = listing.get("hdpData", {}).get("homeInfo", {})
        row = {
            'zpid': listing.get('zpid'),
            'rawHomeStatusCd': listing.get('rawHomeStatusCd'),
            'homeStatusForHDP': hdp_data.get('homeStatusForHDP'),
            'homeStatus': hdp_data.get('homeStatus'),
            'homeType': hdp_data.get('homeType'),
            'address': listing.get('address'),
            'addressZipcode': listing.get('addressZipcode'),
            'addressCity': listing.get('addressCity'),
            'addressState': listing.get('addressState'),
            'latitude': hdp_data.get('latitude', listing.get('latLong', {}).get('latitude')),
            'longitude': hdp_data.get('longitude', listing.get('latLong', {}).get('longitude')),
            'detailUrl': listing.get('detailUrl'),
            'unformattedPrice': listing.get('unformattedPrice'),
            'beds': listing.get('beds'),
            'baths': listing.get('baths'),
            'area': listing.get('area'),
            'zestimate': listing.get('zestimate'),
            'rentZestimate': hdp_data.get('rentZestimate'),
            'taxAssessedValue': hdp_data.get('taxAssessedValue'),
            'timeOnZillow': hdp_data.get('timeOnZillow'),
            'city_tag': listing.get("city_tag")  # Add city tag to the CSV row
        }
        writer.writerow(row)

print(f"Unique listings from all cities saved to '{file_name}'")


```
